# Numpy

```python
# array of numbers 0-100 (exclusive)
list_nums = np.arange(stop=100)
print(list_nums)
```

```python
# list of even numbers from 0-100 (exclusive)
list_even_nums = np.arange(0, 100, 2).tolist()
print(list_even_nums)
```

```python
# accessing elements
print(list_nums[0])
# accessing elements across a range or from the start/end
print(list_nums[5:15])
print(list_nums[:15])
# accessing from the back
print(list_nums[-1])
```

```python
# print element at 90th index
print(list_nums[90])
# elements from index 10-21
print(list_nums[10:22])
# last 10 elements of the array
print(list_nums[-10:])
```

```python
# math operations on lists
# add 1 to each element in the array
list_nums = list_nums + 1
print(list_nums)
```

```python
# natural log of each element in the array
print(np.log(list_nums))
```

```python
# boolean indexing element-wise across an array
print(list_nums % 2 == 0)
list_even_nums2 = list_nums[list_nums % 2 == 0]
print(list_even_nums2)
```

```python
# loop
list_loop = np.array([]) ## empty array
for i in range(0, 101):
    if (i % 2 == 0):
        list_loop = np.append(list_loop, i)

print(list_loop)
```

```python
# list comprehension
list_comp = np.arange(200)
list_comp = np.array([2*i for i in list_comp if i <= 50])
print(list_comp)
```

# Dataframes + Pandas

```python
gapminder = pd.read_csv("data/gapminder.csv")
```

```python
# indexing dataframes by rows
print(gapminder.iloc[1])
```

```python
print(gapminder.loc[1])
# loc = label-based, iloc = index-based
```

```python
# print a range of rows
print(gapminder.iloc[0:5])
```

```python
# printing columns
print(gapminder.loc[:,'continent'])
```

```python
# another way to print cols
print(gapminder['continent'])
```

```python
# printing multiple cols
print(gapminder[['continent', 'country']])
```

```python
# filtering data frames
filtered_gapminder = gapminder[(gapminder['year'] >= 1980) & (gapminder['year'] <= 2000)]
filtered_gapminder
```
